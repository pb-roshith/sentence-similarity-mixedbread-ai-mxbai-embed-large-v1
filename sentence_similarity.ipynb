```python
from transformers import AutoTokenizer, AutoModel
import torch
import torch.nn.functional as F
```

```python
tokenizer = AutoTokenizer.from_pretrained("mixedbread-ai/mxbai-embed-large-v1")
model = AutoModel.from_pretrained("mixedbread-ai/mxbai-embed-large-v1")
```

```python
def sentence_similarity(sent1, sent2):
    # Tokenize sentences
    inputs1 = tokenizer(sent1, return_tensors="pt", truncation=True, padding=True)
    inputs2 = tokenizer(sent2, return_tensors="pt", truncation=True, padding=True)

    # Generate embeddings
    with torch.no_grad():
        embed1 = model(**inputs1).last_hidden_state.mean(dim=1)
        embed2 = model(**inputs2).last_hidden_state.mean(dim=1)

    # Compute cosine similarity
    similarity = F.cosine_similarity(embed1, embed2).item()
    return similarity
```

```python
sentence_similarity("AI is transforming the world.", "Artificial intelligence is changing everything.")
```

```python
sentence_similarity("india is big.", "i play cricket.")
```

```python
sentence_similarity("india is big.", "new movie released in usa.")
```

```python
sentence_similarity("i code well.", "i get a job.")
```

```python
sentence_similarity("i play cricket very well.", "i won trophy in cricket.")
```
